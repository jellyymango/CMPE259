```python
# importing required libraries
import re
import nltk
from nltk.corpus import gutenberg
import string
import pandas as pd
from nltk.corpus import stopwords
nltk.download('gutenberg')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
words = gutenberg.words()
len(words)
```

```python
pattern = r'''(?x) # set flag to allow verbose regexps
(?:[A-Z]\.)+
 | \w+(?:-\w+)* # abbreviations, e.g. U.S.A. # words with optional internal hyphens
 | \$?\d+(?:\.\d+)?%? # currency, percentages, e.g. $12.40, 82%
 | \.\.\. # ellipsis
 | [][.,;"'?():_-] # these are separate tokens; includes ], [
 '''
tokens = re.findall(pattern, gutenberg.raw())
len(tokens)
```

```python
unique_words = set(words)
len(unique_words)
```

```python
unique_words = set(tokens)
len(unique_words)
```

```python
# random words
import random
rand1 = random.randint(300,350)
rand2 = random.randint(360,382)
random_words = tokens[rand1:rand2]
random_words
```

```python
mysentence = "N.L.P. i s a super-easy concept to try..."
mytokens = re.findall(pattern, mysentence)
mytokens
```

```python
from nltk.stem import PorterStemmer
# Initialize the Porter Stemmer
stemmer = PorterStemmer()
# Apply stemming
stemmed_words = [stemmer.stem(word) for word in random_words]
for word, stem_word in zip(random_words[:10], stemmed_words[:10]):
  print("Original: ", word, " -> ", "Stemmed: ", stem_word)
```

```python
stemmer.stem('universe')
```

```python
from nltk import WordNetLemmatizer
# Download necessary resources
nltk.download('punkt')
nltk.download('wordnet')
# Initialize the WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()
# Apply lemmatization
lemmatized_words = [lemmatizer.lemmatize(word) for word in random_words]
# Display the first few lemmatized words
for word, lem_word in zip(random_words[:10], lemmatized_words[:10]):
  print("Original: ", word, " -> ", "Lemmetized: ", lem_word)
```
