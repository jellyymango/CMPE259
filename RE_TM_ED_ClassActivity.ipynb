### Regular Expression Operators and Symbols

| Symbol | Description | Example / Note |
| :--- | :--- | :--- |
| **Basic Matches & Wildcards** | | |
| `.` | **Wildcard**: Matches any single character (except a newline),. | `r"beg.n"` matches *begun*, *begin*, *began*. |
| `\` | **Escape Character**: Used to treat special characters (like `.`, `*`, `?`) literally,. | `\.` matches a period; `\?` matches a question mark. |
| **Disjunction & Ranges** | | |
| `[]` | **Character Disjunction**: Matches any one character inside the brackets. | `[mM]` matches *m* or *M*. |
| `-` | **Range**: Used inside brackets to specify a range of characters. | `` matches a digit; `[a-z]` matches a lower case letter. |
| `^` | **Negation**: When used as the first symbol in `[]`, it matches any character *except* those listed. | `[^a]` matches any character except *a*. |
| `|` | **Disjunction (Pipe)**: Matches either the pattern on the left or the right,. | `cat|dog` matches *cat* or *dog*. |
| **Counters (Quantifiers)** | | |
| `?` | **Question Mark**: Matches zero or one occurrence of the previous character or group. | `colou?r` matches *color* or *colour*. |
| `*` | **Kleene Star**: Matches zero or more occurrences of the previous character or group. | `a*` matches *a*, *aa*, or an empty string. |
| `+` | **Kleene Plus**: Matches one or more occurrences of the previous character or group. | `+` matches a sequence of digits. |
| `{n}` | **Exact Count**: Matches exactly *n* occurrences of the previous character or group. | `x{10}` matches ten *x*s. |
| `{n,m}` | **Range Count**: Matches from *n* to *m* occurrences (mentioned in summary). | |
| `*?` | **Non-greedy Star**: Matches as little text as possible (zero or more). | |
| `+?` | **Non-greedy Plus**: Matches as little text as possible (one or more). | |
| **Anchors & Boundaries** | | |
| `^` | **Start of Line**: Matches the start of a line. | `^The` matches *The* only at the start of a line. |
| `$` | **End of Line**: Matches the end of a line. | `dog.$` matches *dog.* at the end of a line. |
| `\b` | **Word Boundary**: Matches a boundary between a word and a non-word character. | `\bthe\b` matches *the* but not *other*. |
| `\B` | **Non-Word Boundary**: Matches a position that is not a word boundary. | |
| **Groups & Lookaheads** | | |
| `()` | **Grouping/Capture**: Groups terms for order of operations and captures the match for use in substitution,. | `(Column+ +)*` matches repeated columns. |
| `\1`, `\2` | **Backreference**: Refers to the contents of the *n*th captured group. | `\1` refers to the first captured group. |
| `(?:)` | **Non-Capturing Group**: Groups terms without capturing/storing the match,. | `(?: pattern)`. |
| `(?=)` | **Positive Lookahead**: Matches if the pattern occurs next, but does not advance the pointer (zero-width),. | `(?= pattern)`. |
| `(?!)` | **Negative Lookahead**: Matches if the pattern does *not* occur next (zero-width),. | `(?! pattern)`. |
| **Character Class Aliases** | | |
| `\d` | **Digit**: Equivalent to ``. | Matches any digit. |
| `\D` | **Non-Digit**: Equivalent to `[^0-9]`. | Matches any non-digit. |
| `\w` | **Alphanumeric**: Equivalent to `[a-zA-Z0-9_]`. | Matches any alphanumeric character or underscore. |
| `\W` | **Non-Alphanumeric**: Equivalent to `[^\w]`. | Matches symbols like `!`. |
| `\s` | **Whitespace**: Equivalent to `[ \r\t\n\f]`. | Matches space, tab, newline, etc.. |
| `\S` | **Non-Whitespace**: Equivalent to `[^\s]`. | Matches visible characters. |
| `\n` | **Newline**: Matches a newline character. | |
| `\t` | **Tab**: Matches a tab character. | |
| **Unicode Properties** | *(Specifically for the Python `regex` library)* | |
| `\p{L}` | **Unicode Letter**: Matches any Unicode letter,. | |
| `\P{L}` | **Non-Letter**: Matches any Unicode non-letter,. | |
| `\p{N}` | **Number**: Matches any Unicode number,. | |
| `\P{N}` | **Non-Number**: Matches any Unicode non-number,. | |

### Operator Precedence
The sources also specify a hierarchy of precedence for these operators, from highest to lowest:
1.  **Parenthesis**: `()`
2.  **Counters**: `*`, `+`, `?`, `{}`
3.  **Sequences and Anchors**: `the`, `^my`, `end$`
4.  **Disjunction**: `|`

```python
# importing required libraries
import re
import nltk
from nltk.corpus import movie_reviews
import string
import pandas as pd
from nltk.corpus import stopwords
```

```python
sentence = 'I am 20 years old. My previous license plate number was 4XUI302 and my new one is 3A-278. My address is 123 Main street, San Jose, CA. Please call me at (408)-924-1000'
```

```python
# The format of license plate number is a digit then 2 or 3 letters (one of which can be a "-"), and then 3 digits

regex = re.compile(r'(\d[A-Za-z-]{2,3}\d{3})')
lincense_plate_numbers = regex.findall(sentence)
lincense_plate_numbers
```

```python
test_licence_number = "3---278"
result = regex.findall(test_licence_number)
result
```
